```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load dataset
df = pd.read_csv("/content/Global YouTube Statistics.csv", encoding='latin-1')
```

 # **Univariate Analysis**

```python
# Distribution of Subscribers
plt.figure(figsize=(5,3))
sns.histplot(df['subscribers'], kde=True)
plt.title("Distribution of Subscribers")
plt.show()
```

**Insight**

- Most channels have low subscriber counts.

- Very few channels have extremely high subscribers.

- Distribution is right-skewed.


```python
# Distribution of Views
plt.figure()
sns.histplot(df['video views'], kde=True)
plt.title("Distribution of Video Views")
plt.show()
```

**Insight**

- Views show heavy skewness.

- Few viral channels dominate total views.

```python
# Distribution of Earnings
plt.figure()
sns.boxplot(x=df['highest_yearly_earnings'])
plt.title("Distribution of Earnings")
plt.show()
```

**Insight**

- Strong presence of outliers.

- Few channels generate extremely high earnings.

```python
# Country Distribution
plt.figure(figsize=(5,3))
df['Country'].value_counts().head(10).plot(kind='bar')
plt.title("Top Countries by Channel Count")
plt.show()
```

**Insights**
- United States has the highest channel count by a large margin compared to other countries.

- India ranks second, but its count is almost half of the United States.

- There is a sharp drop after the top two countries, with Brazil and the United Kingdom having moderate counts.

- The remaining countries (Mexico, Indonesia, Spain, Thailand, South Korea, Russia) have relatively lower and similar channel counts.

```python
# Category Distribution
plt.figure(figsize=(5,3))
df['category'].value_counts().plot(kind='bar')
plt.title("Distribution of Channel Categories")
plt.show()
```

**Insights**

- Entertainment has the highest number of channels, followed by Music, showing strong audience preference for these categories.

- There is a significant drop in channel count after the top three categories (Entertainment, Music, People & Blogs).

- Mid-level categories like Gaming, Comedy, Education, and Howto & Style have moderate representation.

- Categories such as Travel & Events, Autos & Vehicles, and Nonprofits & Activism have the lowest channel counts, indicating niche presence.

```python
#Channel Category Distribution
plt.figure(figsize=(5,5))
df['category'].value_counts().head(6).plot(kind='pie',autopct='%1.1f%%')
plt.title("Top Channel Categories Distribution")
plt.ylabel("")
plt.show()
```

**Insights**

- Entertainment holds the largest share (30.7%), making it the dominant category.

- Music follows closely with 25.8%, together with Entertainment forming more than half of the total distribution.

- People & Blogs (16.8%) and Gaming (12.0%) have moderate contributions.

- Comedy (8.8%) and Film & Animation (5.9%) represent the smallest shares among the top categories

```python
#Country Distribution
plt.figure(figsize=(5,5))
df['Country'].value_counts().head(6).plot(kind='pie',autopct='%1.1f%%')
plt.title("Top Countries by Channel Contribution")
plt.ylabel("")
plt.show()
```

**Insights**


- The United States contributes the largest share at 48.4 percent, accounting for nearly half of the total channel contribution.

- India holds the second position with 26.0 percent, showing a strong but significantly lower share than the United States.

- Brazil contributes 9.6 percent, while the United Kingdom accounts for 6.6 percent, indicating moderate contributions.

- Mexico and Indonesia contribute smaller shares at 5.1 percent and 4.3 percent respectively.

#**Bivariate Analysis**

```python
#Subscribers vs Views
plt.figure(figsize=(5,3))
sns.scatterplot(x='subscribers', y='video views', data=df)
plt.title("Subscribers vs Views")
plt.show()
```

**Insight**

- Strong positive relationship.

- Channels with more subscribers generally gain more views.

```python
#Earnings vs Subscribers
plt.figure(figsize=(5,3))
sns.scatterplot(x='subscribers', y='highest_yearly_earnings', data=df)
plt.title("Subscribers vs Earnings")
plt.show()
```

**Insight**

- Higher subscribers often lead to higher earnings.

- However, some channels earn differently due to monetization strategies.

```python
#Category vs Average Views
category_views = df.groupby('category')['video views'].mean().sort_values(ascending=False)
plt.figure(figsize=(5,3))
category_views.plot(kind='bar')
plt.title("Average Views by Category")
plt.show()
```

**Insight**

- Shows has the highest average views, significantly higher than all other categories.

- Trailers, Education, and Music also have high average views, forming the next top-performing group.

- Mid-level categories include Sports, Film & Animation, Pets & Animals, and Entertainment with moderate average views.

- Travel & Events and Science & Technology record the lowest average views among all categories.

```python
#Country vs Number of Channels
plt.figure(figsize=(5,3))
sns.countplot(y='Country', data=df, order=df['Country'].value_counts().index[:10])
plt.title("Top Countries by Channel Count")
plt.show()
```

**Insights**
- The United States has the highest channel count, clearly leading by a large margin.

- India ranks second but has nearly half the count of the United States.

- Brazil holds the third position, followed by the United Kingdom and Mexico with moderate counts.

- Spain, Thailand, South Korea, and Russia have relatively low channel counts compared to the top countries.

```python
# Uploads vs Subscribers
plt.figure(figsize=(5,3))
sns.scatterplot(x='uploads', y='subscribers', data=df)
plt.title("Uploads vs Subscribers")
plt.show()
```

**Insights**
- There is no strong linear relationship between uploads and subscribers, as high uploads do not always result in high subscriber counts.

- Some channels with relatively low uploads have very high subscribers, indicating content quality or popularity matters more than quantity.

- A few outliers have extremely high subscriber counts compared to others.

- Many channels with very high upload counts have moderate or low subscribers, suggesting diminishing returns from excessive uploads.

#**Multivariate Analysis**

```python
# Heatmap
plt.figure(figsize=(10,7))
sns.heatmap(df.select_dtypes(include=np.number).corr(), annot=True)
plt.title("Heatmap")
plt.show()
```

**Insights**
- Subscribers and video views show a strong positive correlation, indicating that channels with more subscribers generally have higher total views.

- Monthly and yearly earnings are highly positively correlated with video views and subscribers, showing revenue depends strongly on audience size and engagement.

- Uploads have weak correlation with subscribers and views, suggesting quantity of content does not directly determine popularity.

- Country-related factors such as population and urban population show some moderate relationships, while geographic factors like latitude and longitude have very weak correlation with channel performance.

```python
#Pairplot Analysis
sns.pairplot(df[['subscribers','video views','uploads','highest_yearly_earnings']])
plt.show()
```

**Insights**
- **Subscribers and video views show a positive trend →** channels with more subscribers generally have higher total views.


- **Highest yearly earnings slightly increase with subscribers and views**, but the relationship is not very strong (many outliers).


- **Uploads do not show a clear correlation with earnings or subscribers →** posting more videos doesn’t always mean more success.


- **Data is highly skewed →** most channels have low values, while a few channels dominate with extremely high subscribers, views, and earnings.



```python
# Earnings by Category
plt.figure(figsize=(5,3))
sns.boxplot(x='category', y='highest_yearly_earnings', data=df)
plt.xticks(rotation=90)
plt.title("Earnings by Category")
plt.show()
```

**Insights**
- Music, Entertainment, and People & Blogs show higher earning ranges, with many top outliers earning significantly more.

- Most categories have low median earnings, meaning only a few channels make very high income.

- Autos & Vehicles appears to have relatively consistent and higher median earnings compared to many other categories.

- Travel & Events and Nonprofits & Activism show comparatively lower earnings overall.

```python
# Subscribers by Country
plt.figure(figsize=(9,5))
sns.boxplot(x='Country', y='subscribers', data=df)
plt.xticks(rotation=90)
plt.title("Subscribers Distribution by Country")
plt.show()
```

**Insights**

- India and the United States show the highest subscriber counts, with many channels having very large audiences.

- Most countries have similar median subscriber levels, but a few channels create very high outliers.

- Countries like Pakistan and Japan show relatively strong subscriber distributions compared to several others.

- Several smaller countries have lower and more consistent subscriber counts with fewer extreme outliers.

```python
# Views vs Earnings Colored by Category
plt.figure(figsize=(8,8))
sns.scatterplot(x='video views', y='highest_yearly_earnings', hue='category', data=df)
plt.title("Views vs Earnings by Category")
plt.show()
```

**Insights**

- Higher video views generally relate to higher yearly earnings, showing a positive overall trend.

- Entertainment, Music, and Shows categories contain many of the highest earning channels.

- Some channels achieve high views but comparatively lower earnings, showing monetization differences across categories.

- Most channels cluster at lower views and earnings, with only a few extreme high-performing outliers.

#**Overall Conclusion**

• Subscriber count is strongly linked with channel success, influencing both views and earnings.

• The distribution of subscribers, views, and earnings is highly uneven, where a small number of channels dominate the platform.

• Entertainment and Music categories attract the highest audience engagement and popularity.

• YouTube channel creation is concentrated in specific countries, indicating regional dominance in content production.

• Higher video views generally lead to higher earnings, but earnings also depend on monetization methods.

• Upload frequency alone does not guarantee subscriber growth or higher engagement.

• Several outliers exist in views and earnings, representing highly successful or viral channels.

• Channel growth depends on multiple factors including content quality, audience engagement, and niche selection.

# **Key Recommendation**
1. Focus on High-Demand Content Categories

- Creators should consider producing content in popular categories such as entertainment, music, and lifestyle, as these areas show higher engagement and earning potential.

2. Prioritize Subscriber Growth Strategies

- Since subscribers strongly influence views and revenue, creators should invest in audience engagement techniques such as consistent posting schedules, audience interaction, and community building.

3. Improve Content Quality Over Quantity

- The analysis showed that higher upload frequency does not always lead to increased subscribers. Creators should focus on producing high-quality and engaging content rather than simply increasing upload volume.

4. Explore Multiple Revenue Streams

- Some channels with similar view counts generate different earnings levels. Creators should diversify income through sponsorships, merchandise, and collaborations to maximize revenue.

5. Target Region-Specific Audiences

- Country-level trends suggest that regional content and localization strategies can improve audience reach and engagement.

6. Utilize Data Analytics for Channel Growth

- Creators and marketers should continuously analyze performance metrics such as subscriber growth, audience retention, and engagement rates to optimize content strategies.

7. Learn From High-Performing Channels

- Studying successful channels can help identify effective content styles, posting frequency, and branding strategies.
